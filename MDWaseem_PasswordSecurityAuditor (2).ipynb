# Password Security Auditor
A beginner-friendly cybersecurity project that evaluates password hygiene locally. It does not save or transmit the password. Use only test passwords, never share a real password.

## Objectives
- Check length and character variety.
- Flag a small list of common example passwords.
- Give practical improvement tips.

**Limitation:** this is an educational rule-based checker, not a password-cracking tool and not a guarantee that a password is secure.

```python
import re

COMMON_PASSWORDS = {
    'password', 'password123', '12345678', 'qwerty123',
    'admin123', 'iloveyou', 'welcome123', 'letmein'
}

def check_password_strength(password: str) -> dict:
    checks = {
        'At least 12 characters': len(password) >= 12,
        'Contains lowercase letters': bool(re.search(r'[a-z]', password)),
        'Contains uppercase letters': bool(re.search(r'[A-Z]', password)),
        'Contains digits': bool(re.search(r'\d', password)),
        'Contains symbols': bool(re.search(r'[^A-Za-z0-9]', password)),
        'Not a common example password': password.lower() not in COMMON_PASSWORDS,
    }
    score = sum(checks.values())
    if password.lower() in COMMON_PASSWORDS or len(password) < 8:
        rating = 'Weak'
    elif score >= 5 and len(password) >= 12:
        rating = 'Strong'
    elif score >= 4:
        rating = 'Moderate'
    else:
        rating = 'Weak'
    tips = []
    if len(password) < 12: tips.append('Use at least 12 characters.')
    if not re.search(r'[a-z]', password): tips.append('Add lowercase letters.')
    if not re.search(r'[A-Z]', password): tips.append('Add uppercase letters.')
    if not re.search(r'\d', password): tips.append('Add numbers.')
    if not re.search(r'[^A-Za-z0-9]', password): tips.append('Add symbols.')
    if password.lower() in COMMON_PASSWORDS: tips.append('Avoid common or predictable passwords.')
    if not tips: tips.append('Use a unique password for every account and enable MFA.')
    return {'rating': rating, 'score': score, 'checks': checks, 'tips': tips}

print('Function ready. Use a TEST password in the next cell.')
```

```python
test_password = 'Example-Test-928!'
result = check_password_strength(test_password)
print(f"Strength: {result['rating']} ({result['score']}/6 checks passed)")
for label, passed in result['checks'].items():
    print(f"[{'PASS' if passed else 'FAIL'}] {label}")
print('\nSuggestions:')
for tip in result['tips']:
    print('-', tip)
```
